## Shape Error

Another important error is shape error. The two tensors must have in same inner shapes.

example:
***
1. The inner dimensions must match:

~~~
(3, 2) @ (3, 2) won't work
(2, 3) @ (3, 2) will work
(3, 2) @ (2, 3) will work
~~~
2. The resulting matrix has the shape of the outer dimensions:

~~~
(2, 3) @ (3, 2) -> (2, 2)
(3, 2) @ (2, 3) -> (3, 3)
~~~

````python
torch.manual_seed(42)
try:
    torch.rand(size=(3, 2)) @ torch.rand(size=(3, 2))
except Exception as e:
    print(e)
````

````python
torch.manual_seed(42)
mat1 = torch.rand(size=(2, 3))
mat2 = torch.rand(size=(3, 2))

print(f"Matrix Multiplication : {mat1 @ mat2}")
print(f"Matrix Shape : {(mat1 @ mat2).shape}")
````

## Reshaping, stacking, squeezing and unsqueezing

|Method|Description|
|---|---|
|`torch.reshape(input, shape)`|Reshapes `input` to `shape`, if also used `torch.Tensor.reshape()`.|
|`torch.view()`|Returns a view of the original tensor in a different shape but shares the same data as the original tensor.|
|`torch.stack(tensors, dim=0)`|Concatenates a sequence of `tensors` along a new dimension ( `dim` ), all `tensors` must be same size.|
|`torch.squeeze(input)`|Squeezes `input` to remove all the dimenions with value1.|
|`torch.unsqueeze(input, dim)`|Returns `input` with a dimension value of 1 added at `dim`.|
|`torch.permute(input, dim)`|Returns a view of the original `input` with its dimensions permuted (rearranged) to `dims` .|

### Reshape

````python
x = torch.arange(1, 7)
print(x)
print(x.shape)

print('-------')
print(x.reshape(3, 2))
print(x.reshape(3, 2).shape)

print('-------')
print(x.reshape(2, 3))
print(x.reshape(3, 2).shape)
````

### View

````python
print('original x', x)
print('viewd x', x.view(3,2))
````

### Stack

````python
a = torch.tensor([2, 4])
b = torch.tensor([5, 6])

print(torch.stack([a, b], dim=0))
print(torch.stack([a, b], dim=1))
````

### Squeeze

````python
a2 = torch.rand(size=(1, 3, 3))
print(a2)
print(a2.shape)
print(a2.ndim)
print('-'*40)

print(a2.squeeze())
print(a2.squeeze().shape)
print(a2.squeeze().ndim)
print('-'*40)
````

### Unsqueeze

````python
a3 = torch.rand(size=(1, 3, 3))
print(a3)
print(a3.shape)
print(a3.ndim)
print('-'*40)

print(a3.unsqueeze(dim=1))
print(a3.unsqueeze(dim=1).shape)
print(a3.unsqueeze(dim=1).ndim)
````

### Permute

````python
xtensor = torch.rand(size=(2, 3, 3))
print(xtensor.shape)
````

````python
x = xtensor.permute(2, 1,0 )
print(x.shape)
````

````python
x = xtensor.permute(1,0,2 )
print(x.shape)
````
