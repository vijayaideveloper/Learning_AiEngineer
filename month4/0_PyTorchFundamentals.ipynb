## shapes and dimensions

In pytorch mismatched the shapes and dimensions while matrix maultiplications. which is caused an error.

```python
# scaler is the 0 dimension

scaler = torch.tensor(7)
print(f'Scaler : {scaler}')
print(f"scaler shape : {scaler.shape}")
print(f"scaler dimension : {scaler.ndim}")
print(f"scaler item : {scaler.item()}")
```

```python
# vector is the series of numbers

vector = torch.tensor([1,2,3,4,5,6,7,8])
print(f"vector : {vector}")
print(f'vector shape : {vector.shape}')
print(f'vector dimension : {vector.ndim}')
```

```python
matrix = torch.tensor([
    [2,3,4],
    [5,3,7],
    [5,7,8]
])

print(f"matrix : {matrix}")
print(f"matrix shape : {matrix.shape}")
print(f"matrix dimensions : {matrix.ndim}")
```
