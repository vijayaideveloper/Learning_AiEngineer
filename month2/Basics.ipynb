## Mysql

`Mysql` is the RDBMS - relational database management system. which is used to manage datas.

which is represented in rows and columns.mysql is open source. most of the syntax is matched to other databases.

## SQL Querry in RDBMS

1. Create table

Syntax:
~~~
CREATE TABLE table_name(
    column1_name DATATYPE constraint,
    column1_name DATATYPE constraint
);
~~~
***
2. Insert into Table

Syntax:

~~~
INSERT INTO table_name(column1_name, column2_name,..) VALUES (value1, value1,..);

or 

INSERT INTO table_name VALUES (,...);
~~~
***
3. Querying Data (SELECT)

Syntax:

~~~
SELECT column1, column2,..
FROM table_name
WHERE contition;
~~~
***
4. Deleting Data data from Table

Syntax:

~~~
DELETE FROM table_name
WHERE contition;
~~~
***
5. Drop a Table

Syntax:

~~~
DROP TABLE table_name;
~~~
***
